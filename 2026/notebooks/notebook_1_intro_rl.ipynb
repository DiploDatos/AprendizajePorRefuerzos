### Repaso rápido


* **Recompensa:** señal $r$ que recibe el agente después de cada acción.

* **Valor:** $V_\pi(s)$ es la recompensa futura esperada desde $s$ al seguir $\pi$; $Q_\pi(s,a)$ es la recompensa futura esperada al tomar $a$ en $s$ y continuar con $\pi$.

* **Política:** indica cómo elegir acciones. En una política $\epsilon$-greedy, con probabilidad $1-\epsilon$ se elige una acción con el mayor valor Q y con probabilidad $\epsilon$ se elige una acción al azar. Si hay una única acción óptima entre $|A|$ acciones, su probabilidad total es $1-\epsilon+\epsilon/|A|$; la de cada otra acción es $\epsilon/|A|$.

* **Softmax:** asigna una probabilidad a cada acción según su valor y luego realiza un sorteo: $\pi(a\mid s)=\frac{e^{Q(s,a)/\tau}}{\sum_{b\in A}e^{Q(s,b)/\tau}}$. El parámetro $\tau>0$ controla cuán concentrada está la elección en las acciones más valiosas.

En este notebook vemos dos algoritmos para actualizar la función de valor (y, por lo tanto, la política de selección de acciones):

* Actualización por SARSA (on-policy).

$$Q(s,a) \gets Q(s,a) + \alpha (r + \gamma Q(s',a') - Q(s,a))$$

Algoritmo completo (a modo de referencia):

![Algoritmo SARSA](https://github.com/DiploDatos/AprendizajePorRefuerzos/blob/master/images/sarsa.png?raw=1)


* Actualización por Q-Learning (*off-policy*):

$$Q(s,a) \gets Q(s,a) + \alpha\left[r + \gamma \max_{a'} Q(s',a') - Q(s,a)\right]$$

Si $s'$ es terminal, el valor futuro del objetivo es cero.

Algoritmo completo (a modo de referencia):

![Algoritmo Q-Learning](https://github.com/DiploDatos/AprendizajePorRefuerzos/blob/master/images/q_learning.png?raw=1)

Fuente de las imágenes: capítulo 6 de [Reinforcement Learning: An Introduction](http://www.incompleteideas.net/book/the-book.html).

```python
# @title Gráficos de entrenamiento y tabla Q
import numpy as np
import matplotlib.pyplot as plt


def _plot_series(values, title, ylabel, window=50):
    values = np.asarray(values, dtype=float)
    episodes = np.arange(1, len(values) + 1)
    plt.figure(figsize=(9, 4))
    plt.plot(episodes, values, alpha=0.3, label="Por episodio")
    if len(values):
        width = min(window, len(values))
        smooth = np.convolve(values, np.ones(width) / width, mode="valid")
        plt.plot(episodes[width - 1:], smooth, label=f"Media móvil ({width})")
    plt.xlabel("Episodio")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.legend()
    plt.show()


def plot_reward_per_episode(reward_ep):
    _plot_series(reward_ep, "Recompensa durante el entrenamiento", "Recompensa")


def plot_steps_per_episode(timesteps_ep):
    _plot_series(timesteps_ep, "Pasos durante el entrenamiento", "Pasos")


def draw_value_matrix(q):
    rows, cols = 4, 12
    values = np.full((rows, cols), np.nan)
    arrows = ("↑", "→", "↓", "←")
    labels = np.full((rows, cols), "?", dtype=object)

    for row in range(rows):
        for col in range(cols):
            state = row * cols + col
            if row == 3 and 1 <= col <= 10:
                labels[row, col] = "C"
            elif state == 47:
                labels[row, col] = "G"
            elif any((state, action) in q for action in range(4)):
                action_values = [q.get((state, action), 0.0) for action in range(4)]
                values[row, col] = max(action_values)
                prefix = "S " if state == 36 else ""
                labels[row, col] = f"{prefix}{arrows[np.argmax(action_values)]}\n{values[row, col]:.1f}"
            elif state == 36:
                labels[row, col] = "S ?"

    cmap = plt.colormaps["RdYlGn"].copy()
    cmap.set_bad("#eeeeee")
    fig, ax = plt.subplots(figsize=(14, 5))
    image = ax.imshow(np.ma.masked_invalid(values), cmap=cmap)
    for row in range(rows):
        for col in range(cols):
            ax.text(col, row, labels[row, col], ha="center", va="center", fontsize=9)
    ax.set_xticks(range(cols))
    ax.set_yticks(range(rows))
    ax.set_title("Mejor valor Q y acción por estado; ? = sin visitar, C = acantilado")
    fig.colorbar(image, ax=ax, label="máx Q(s, a)")
    fig.tight_layout()
    plt.show()
```

Definimos el método de elección de acciones. En este caso el mismo utiliza la política de exploración $\epsilon$-greedy.

```python
def choose_action_e_greedy(
        state,
        actions: range,
        q: dict,
        hyperparameters: dict,
        random_state: np.random.RandomState,
) -> int:
    """
    Elige una acción de acuerdo al aprendizaje realizado previamente
    usando una política de exploración épsilon-greedy
    """
    # ej: para 4 acciones inicializa en [0,0,0,0]
    q_values = [q.get((state, a), 0.0) for a in actions]
    max_q = max(q_values)
    # sorteamos un número: es menor a épsilon?
    if random_state.uniform() < hyperparameters['epsilon']:
        # sí: se selecciona una acción aleatoria
        return random_state.choice(actions)

    count = q_values.count(max_q)

    # hay más de un máximo valor de estado-acción?
    if count > 1:
        # sí: seleccionamos uno de ellos aleatoriamente
        best = [i for i in range(len(actions)) if q_values[i] == max_q]
        i = random_state.choice(best)
    else:
        # no: seleccionamos el máximo valor de estado-acción
        i = q_values.index(max_q)

    return actions[i]
```

Finalmente, definimos el método principal de iteraciones.

```python
def run(learning_function, hyperparameters, episodes_to_run, env, actions,
        q, random_state, max_steps=2000, choose_action=choose_action_e_greedy,
        state_encoder=None):
    """Entrena un agente y registra pasos y recompensa de cada episodio."""
    if state_encoder is None:
        state_encoder = lambda observation: observation
    timesteps_of_episode = []
    reward_of_episode = []

    for episode in range(episodes_to_run):
        observation, _ = env.reset(seed=episode)
        state = state_encoder(observation)
        action = choose_action(state, actions, q, hyperparameters, random_state)
        episode_reward = 0.0

        for step in range(max_steps):
            next_observation, reward, terminated, truncated, _ = env.step(action)
            next_state = state_encoder(next_observation)
            episode_reward += reward
            next_action = None if terminated else choose_action(
                next_state, actions, q, hyperparameters, random_state
            )
            learning_function(
                state, action, reward, next_state, next_action,
                terminated, hyperparameters, q, actions
            )
            if terminated or truncated:
                break
            state, action = next_state, next_action

        timesteps_of_episode.append(step + 1)
        reward_of_episode.append(episode_reward)

    steps = np.asarray(timesteps_of_episode)
    rewards = np.asarray(reward_of_episode)
    return steps.mean(), steps, rewards
```

A diferencia de lo que sucede en el aprendizaje supervisado, en el aprendizaje por refuerzos el rendimiento se evalúa por una función específica que es la función de recompensa. En la práctica, la función de recompensa puede ser externa (y provista por el entorno) o bien puede ser una función creada por diseño (a modo de dirigir el agente hacia lo que por diseño se considera mejor, en nuestro ejemplo podría ser con una recompensa de $+1$ cada vez que el agente llega al estado objetivo). Esto se conoce como *reward shaping*, y hay que tener mucho cuidado con los posibles efectos secundarios de su uso.

Como el objetivo de RL es maximizar la recompensa obtenida, es posible utilizar la información sobre la obtención de la recompensas en cada time-step o episodio para evaluar el rendimiento parcial del agente (esto depende mucho de la particularidad de la distribución de la recompensa para el problema tratado).

## De *The Cliff* a CartPole y MountainCar

En *The Cliff*, cada observación ya es un número entero. CartPole devuelve cuatro valores continuos (posición y velocidad del carro; ángulo y velocidad angular del poste), y MountainCar devuelve dos (posición y velocidad). Para conservar una tabla Q, dividimos cada dimensión en intervalos y usamos la tupla de índices como estado. Este método pierde precisión y el tamaño de la tabla crece rápidamente al agregar intervalos.

En ambos entornos, la recompensa por episodio coincide con los pasos o con su opuesto. Por eso mostramos una curva de recompensa y, aparte, la recompensa de evaluación sin exploración.

Las funciones `learn_SARSA` y `learn_Q_learning` que completaste arriba se reutilizan sin cambios. Primero ejecutá las celdas de *The Cliff* hasta definir esas funciones. Los entrenamientos siguientes pueden tardar algunos minutos, sobre todo en MountainCar.

```python
def make_discretizer(lower, upper, bins):
    """Convierte una observación continua en una tupla de índices."""
    if not (len(lower) == len(upper) == len(bins)):
        raise ValueError("Cada dimensión necesita límites y cantidad de intervalos")
    if any(count < 2 for count in bins):
        raise ValueError("Cada dimensión necesita al menos dos intervalos")
    edges = [np.linspace(lo, hi, count + 1)[1:-1]
             for lo, hi, count in zip(lower, upper, bins)]

    def encode(observation):
        return tuple(int(np.searchsorted(boundaries, value))
                     for boundaries, value in zip(edges, observation))

    return encode


def evaluate_greedy(env_name, q, encoder, episodes=30, seed=10_000):
    """Evalúa la política aprendida sin exploración, en episodios nuevos."""
    env_eval = gym.make(env_name)
    actions_eval = range(env_eval.action_space.n)
    random_state_eval = np.random.RandomState(seed)
    rewards, steps, reached_goal = [], [], []
    greedy_parameters = {"epsilon": 0.0}
    try:
        for episode in range(episodes):
            observation, _ = env_eval.reset(seed=seed + episode)
            total_reward = 0.0
            for timestep in range(1, 2001):
                state = encoder(observation)
                action = choose_action_e_greedy(
                    state, actions_eval, q, greedy_parameters, random_state_eval
                )
                observation, reward, terminated, truncated, _ = env_eval.step(action)
                total_reward += reward
                if terminated or truncated:
                    break
            rewards.append(total_reward)
            steps.append(timestep)
            reached_goal.append(terminated)
    finally:
        env_eval.close()
    return np.asarray(rewards), np.asarray(steps), np.asarray(reached_goal)


def compare_algorithms(env_name, encoder, hyperparameters, episodes, max_steps=2000):
    """Entrena SARSA y Q-Learning desde cero con los mismos parámetros."""
    results = {}
    for name, learner in (("SARSA", learn_SARSA), ("Q-Learning", learn_Q_learning)):
        env_train = gym.make(env_name)
        q_values = {}
        random_state_train = np.random.RandomState(42)
        try:
            _, steps, rewards = run(
                learner, hyperparameters, episodes, env_train,
                range(env_train.action_space.n), q_values, random_state_train,
                max_steps=max_steps, state_encoder=encoder
            )
        finally:
            env_train.close()
        eval_rewards, eval_steps, reached_goal = evaluate_greedy(
            env_name, q_values, encoder
        )
        results[name] = {
            "q": q_values, "train_rewards": rewards, "train_steps": steps,
            "eval_rewards": eval_rewards, "eval_steps": eval_steps,
            "eval_goal": reached_goal,
        }
        print(f"{env_name} · {name}: entrenamiento {rewards[-100:].mean():.1f} "
              f"de recompensa media (últimos 100); evaluación "
              f"{eval_rewards.mean():.1f} ± {eval_rewards.std():.1f} "
              f"en {len(eval_rewards)} episodios")
    return results


def plot_comparison(results, title, window=50):
    """Muestra la curva de entrenamiento y la evaluación sin exploración."""
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    names = list(results)
    for name, result in results.items():
        values = np.asarray(result["train_rewards"], dtype=float)
        width = min(window, len(values))
        smoothed = np.convolve(values, np.ones(width) / width, mode="valid")
        axes[0].plot(np.arange(width, len(values) + 1), smoothed, label=name)
    axes[0].set(xlabel="Episodio", ylabel="Recompensa",
                title=f"{title}: entrenamiento")
    axes[0].legend()

    means = [results[name]["eval_rewards"].mean() for name in names]
    deviations = [results[name]["eval_rewards"].std() for name in names]
    axes[1].bar(names, means, yerr=deviations, capsize=5)
    axes[1].set(ylabel="Recompensa media ± desviación estándar",
                title=f"{title}: evaluación sin exploración")
    fig.tight_layout()
    plt.show()
```

### CartPole

La recompensa aumenta mientras el poste siga equilibrado: el episodio termina al caer o al alcanzar el límite de 500 pasos. Acotamos las velocidades para construir intervalos finitos; los valores fuera de esos límites quedan en el primer o último intervalo. Una recompensa media cercana a 500 en la evaluación indica que el agente suele sostener el poste hasta el límite.

```python
cartpole_encoder = make_discretizer(
    lower=(-2.4, -3.0, -0.21, -3.5),
    upper=(2.4, 3.0, 0.21, 3.5),
    bins=(2, 2, 8, 12),
)
cartpole_parameters = {"alpha": 0.5, "gamma": 0.99, "epsilon": 0.1}
cartpole_results = compare_algorithms(
    "CartPole-v1", cartpole_encoder, cartpole_parameters, episodes=2000
)
plot_comparison(cartpole_results, "CartPole")
```

### MountainCar

El auto recibe una recompensa de −1 por paso y necesita tomar impulso para llegar a la cima. Por eso, una recompensa más alta (menos negativa) suele indicar que alcanzó la meta en menos pasos. También mostramos la proporción de episodios de evaluación que llegaron a la meta; un episodio que termina por el límite de 200 pasos no cuenta como éxito.

El aprendizaje puede ser sensible al número de intervalos y a la exploración. Si no llega a la meta, probá más episodios o distintos valores de `epsilon`, `alpha`, `gamma` y `bins`, manteniendo el mismo protocolo de evaluación.

```python
mountaincar_env = gym.make("MountainCar-v0")
mountaincar_encoder = make_discretizer(
    lower=mountaincar_env.observation_space.low,
    upper=mountaincar_env.observation_space.high,
    bins=(24, 24),
)
mountaincar_env.close()
mountaincar_parameters = {"alpha": 0.25, "gamma": 0.99, "epsilon": 0.15}
mountaincar_results = compare_algorithms(
    "MountainCar-v0", mountaincar_encoder, mountaincar_parameters, episodes=1500
)
plot_comparison(mountaincar_results, "MountainCar")
for name, result in mountaincar_results.items():
    print(f"{name}: llegó a la meta en {result['eval_goal'].mean():.0%} "
          "de los episodios de evaluación")
```
